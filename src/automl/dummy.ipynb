```python
import pandas as pd

# Update the path as needed
X = pd.read_parquet("../../data/yprop_4_1/5/X_train.parquet")

print(f"Shape: {X.shape}\n")

constant_features = []

for col in X.columns:
    # Count unique values (including NaN as a unique value)
    n_unique = X[col].nunique(dropna=False)

    if n_unique == 1:
        constant_features.append(col)
        print(f"{col:25s} --> CONSTANT")
        print(f"    Value: {X[col].iloc[0]}\n")

print("-" * 50)

if len(constant_features) == 0:
    print("No constant features found.")
else:
    print(f"{len(constant_features)} constant feature(s) found:")
    print(constant_features)
```

```python
import pandas as pd

X = pd.read_parquet("../../data/bike_sharing_demand/1/X_train.parquet")   # <- update this path

print(f"shape: {X.shape}\n")

for col in X.columns:
    dtype = X[col].dtype
    n_unique = X[col].nunique(dropna=True)
    sample_vals = X[col].dropna().unique()[:5]

    if dtype.name in ("object", "category"):
        status = "CATEGORICAL / TEXT (needs encoding)"
    elif dtype.kind in "iu" and n_unique <= 20:
        status = "numeric, low-cardinality (already encoded, or just a numeric feature)"
    else:
        status = "numeric (continuous)"

    print(f"{col:25s} dtype={str(dtype):20s} n_unique={n_unique:5d}  {status}")
    print(f"    sample values: {list(sample_vals)}")

print("\n--- SUMMARY ---")
raw_categorical = [c for c in X.columns if X[c].dtype.name in ("object", "category")]
print(f"Columns still needing encoding: {len(raw_categorical)}")
if raw_categorical:
    print(f"  -> {raw_categorical}")
else:
    print("  -> None. All columns are already numeric.")
```
