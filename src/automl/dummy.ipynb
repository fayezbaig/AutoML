```python
import pandas as pd

for fold in range(1, 11):
    Xtr = pd.read_parquet(f"../../data/wine_quality/{fold}/X_train.parquet")
    Xte = pd.read_parquet(f"../../data/wine_quality/{fold}/X_test.parquet")

    # exact duplicate rows shared between train and test
    merged = Xtr.merge(Xte.drop_duplicates(), how="inner")
    print(f"fold {fold}: train={len(Xtr)} test={len(Xte)} exact_overlap={len(merged)}")

    # does any single feature look like a leaking index?
    corrs = Xtr.assign(_row=range(len(Xtr))).corr(numeric_only=True)["_row"].drop("_row")
    print("  top |corr with row order|:", corrs.abs().sort_values(ascending=False).head(3).to_dict())
```

```python
import pandas as pd

for fold in range(1, 11):

    Xtr = pd.read_parquet(f"../../data/wine_quality/{fold}/X_train.parquet")
    Xte = pd.read_parquet(f"../../data/wine_quality/{fold}/X_test.parquet")

    ytr = pd.read_parquet(f"../../data/wine_quality/{fold}/y_train.parquet")
    yte = pd.read_parquet(f"../../data/wine_quality/{fold}/y_test.parquet")

    train_full = Xtr.copy()
    train_full["target"] = ytr.values

    test_full = Xte.copy()
    test_full["target"] = yte.values

    overlap = train_full.merge(
        test_full.drop_duplicates(),
        how="inner"
    )

    print(
        f"Fold {fold}: exact X+y overlap = {len(overlap)}"
    )
```
